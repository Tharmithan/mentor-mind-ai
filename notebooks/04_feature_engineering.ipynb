## 4. Feature importance chart

```python
import json
ranked = json.loads((REPO / "datasets" / "final" / "feature_importance.json").read_text())["ranked_features"]
kept = [r for r in ranked if r["kept"]][:12]

plt.figure(figsize=(8, 5))
plt.barh([r["feature"] for r in kept][::-1], [r["importance_pct"] for r in kept][::-1], color="#8b5cf6")
plt.xlabel("Importance %")
plt.title("Top Features (Day 4)")
plt.tight_layout()
plt.show()
```
