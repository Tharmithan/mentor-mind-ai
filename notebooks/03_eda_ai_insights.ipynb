Libraries: `matplotlib` · `seaborn` · `plotly`

```python
import sys
from pathlib import Path

import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns

REPO = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()
sys.path.insert(0, str(REPO / "datasets" / "scripts"))

from eda_analysis import load_cleaned, generate_insights, run_eda

sns.set_theme(style="whitegrid")
%matplotlib inline
```

## Load cleaned data

```python
df = load_cleaned()
df.head()
```

## Correlation matrix

```python
cols = ["study_hours", "attendance_pct", "wellness_score", "past_failures", "performance_pct"]
sns.heatmap(df[cols].corr(), annot=True, fmt=".2f", cmap="RdYlGn", center=0)
plt.title("Correlation — performance drivers")
plt.show()
```

## Study hours vs marks

```python
sns.regplot(data=df, x="study_hours", y="performance_pct", scatter_kws={"alpha": 0.3})
plt.title("Study Hours vs Marks")
plt.show()
```

## Attendance vs performance

```python
sns.scatterplot(data=df, x="attendance_pct", y="performance_pct", hue="at_risk", palette=["#22d3ee", "#f43f5e"])
plt.title("Attendance vs Performance")
plt.show()
```

## Wellness / sleep proxy vs grades

UCI uses `health` → `wellness_score` (1–5) as a sleep/wellness proxy.

```python
sns.boxplot(data=df, x="wellness_score", y="performance_pct", palette="Purples")
plt.xlabel("Wellness / Sleep Proxy (1–5)")
plt.title("Wellness vs Grades")
plt.show()
```

## Subject weaknesses

```python
df.groupby("subject")["performance_pct"].mean().sort_values().plot(kind="bar", color="#8b5cf6")
plt.ylabel("Avg Performance %")
plt.title("Subject Weaknesses")
plt.xticks(rotation=0)
plt.show()
```
